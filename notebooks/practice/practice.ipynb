```python
import xarray as xr
import numpy as np

darr1 = xr.DataArray(np.array([[1, 2, 3]]), 
    dims=["d1","d2"],
    coords={"d1":["a1"], "d2":["b1", "b2", "b3"]}
)
darr2 = xr.DataArray(np.array([[4, 5], [6, 7]]), 
    dims=["d1","d2"],
    coords={"d1":["a2", "a3"], "d2":["b3", "b4"]}
)
xrs = xr.concat([darr1, darr2], dim="d1", join="outer")
xrs
```
